```python
# Cell 1: setup. The test set is NOT loaded in this cell.
import datetime
import json
import math
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import KFold, StratifiedKFold, cross_val_predict
from sklearn.ensemble import (RandomForestRegressor, GradientBoostingRegressor,
                              RandomForestClassifier, GradientBoostingClassifier)
from sklearn.linear_model import Ridge, LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import (mean_absolute_error, r2_score, roc_auc_score, average_precision_score,
                             confusion_matrix, precision_score, recall_score)

sns.set_theme(style="whitegrid")
train = pd.read_csv("train.csv")
cfg = json.load(open("final_config.json"))

y = train["G3"]
y_risk = (y < 10).astype(int)
X_all = train.drop(columns="G3")
X_early = X_all.drop(columns=["G1", "G2"])
feature_sets = {"early (no G1/G2)": X_early, "all features": X_all}

def build_reg(name, params):
    base = {"Ridge": make_pipeline(StandardScaler(), Ridge()),
            "Random Forest": RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1),
            "Gradient Boosting": GradientBoostingRegressor(random_state=42)}[name]
    return base.set_params(**params)

def build_clf(name, params):
    base = {"Logistic": make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000)),
            "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1),
            "Gradient Boosting": GradientBoostingClassifier(random_state=42)}[name]
    return base.set_params(**params)

rmse = lambda a, b: float(np.sqrt(np.mean((np.asarray(a) - np.asarray(b)) ** 2)))

print("Training students:", len(train), "| at-risk share:", f"{y_risk.mean():.1%}")
print("\nModels fixed on Day 20:")
for fname in feature_sets:
    r, c = cfg["regression"][fname], cfg["classification"][fname]
    print(f"\n{fname}")
    print(f"  regression    : {r['model']:18s} nested RMSE {r['nested_cv_rmse']:.3f} +/- {r['std']:.3f} | useful: {r['useful']}")
    print(f"  classification: {c['model']:18s} nested AUC  {c['nested_roc_auc']:.3f} +/- {c['std']:.3f} | usable: {c['usable']}")
```

```python
# Cell 2: pre-registration. This uses ONLY training data and is saved BEFORE the test set is opened.
rng = np.random.default_rng(0)
n_test = math.ceil(len(train) * 0.25)          # the Day 9 split was 80/20, so the test set is about a quarter of train
kf = KFold(5, shuffle=True, random_state=42)
skf = StratifiedKFold(5, shuffle=True, random_state=42)

pre = {"created": datetime.datetime.now().isoformat(timespec="seconds"),
       "n_test_assumed": n_test, "regression": {}, "classification": {}}

for fname, X in feature_sets.items():
    # regression: out-of-fold predictions, then "what would a sample of n_test students give?"
    info = cfg["regression"][fname]
    p = cross_val_predict(build_reg(info["model"], info["params"]), X, y, cv=kf)
    boot = [rmse(y.values[i], p[i]) for i in (rng.integers(0, len(y), n_test) for _ in range(3000))]
    lo, hi = np.percentile(boot, [2.5, 97.5])
    pre["regression"][fname] = {"model": info["model"], "expected_rmse": info["nested_cv_rmse"],
                                "expected_range_95": [float(lo), float(hi)]}

    # classification: out-of-fold probabilities, expected AUC range, and the FIXED threshold
    c = cfg["classification"][fname]
    pr = cross_val_predict(build_clf(c["model"], c["params"]), X, y_risk, cv=skf, method="predict_proba")[:, 1]
    aucs = []
    for _ in range(3000):
        i = rng.integers(0, len(y_risk), n_test)
        if 0 < y_risk.values[i].sum() < n_test:
            aucs.append(roc_auc_score(y_risk.values[i], pr[i]))
    thr = max(t for t in np.linspace(0.01, 0.99, 99)
              if recall_score(y_risk, (pr >= t).astype(int)) >= 0.80)
    flag = (pr >= thr).astype(int)
    pre["classification"][fname] = {"model": c["model"], "expected_auc": c["nested_roc_auc"],
                                    "expected_range_95": [float(v) for v in np.percentile(aucs, [2.5, 97.5])],
                                    "threshold": float(thr),
                                    "train_oof_recall": float(recall_score(y_risk, flag)),
                                    "train_oof_precision": float(precision_score(y_risk, flag, zero_division=0)),
                                    "train_oof_flagged_share": float(flag.mean())}

json.dump(pre, open("preregistration.json", "w"), indent=2)

print(f"Assumed test size: {n_test} students (about {round(n_test * y_risk.mean())} at risk)\n")
for fname in feature_sets:
    r, c = pre["regression"][fname], pre["classification"][fname]
    print(f"{fname}")
    print(f"  Expected test RMSE : {r['expected_rmse']:.2f}, 95% range for {n_test} students: "
          f"{r['expected_range_95'][0]:.2f} to {r['expected_range_95'][1]:.2f}")
    print(f"  Expected test AUC  : {c['expected_auc']:.2f}, 95% range: "
          f"{c['expected_range_95'][0]:.2f} to {c['expected_range_95'][1]:.2f}")
    print(f"  Fixed threshold    : {c['threshold']:.2f} (on training data it flags {c['train_oof_flagged_share']:.0%} "
          f"of students, recall {c['train_oof_recall']:.2f}, precision {c['train_oof_precision']:.2f})\n")
print("Saved preregistration.json. Now write your own predictions in a text cell (see section 3), THEN run Cell 3.")
```

```python
# Cell 3: THE ONE-TIME EVALUATION, REGRESSION. Run this cell once only.
if os.path.exists("test_evaluation_done.flag"):
    raise RuntimeError("The test set has already been evaluated in this session. "
                       "Do not rerun, because repeated looks make the estimate dishonest.")

test = pd.read_csv("test.csv")
assert list(test.drop(columns="G3").columns) == list(X_all.columns), "test columns differ from train columns"
y_te = test["G3"].values
y_te_risk = (y_te < 10).astype(int)
print("Test students:", len(test), "| at risk:", y_te_risk.sum(), f"({y_te_risk.mean():.1%})")

rng = np.random.default_rng(1)
idx_boot = [rng.integers(0, len(y_te), len(y_te)) for _ in range(3000)]
rows, test_pred, fitted = [], {}, {}

for fname, Xtr in feature_sets.items():
    info = cfg["regression"][fname]
    model = build_reg(info["model"], info["params"]).fit(Xtr, y)
    p = model.predict(test[Xtr.columns])
    b = np.full(len(y_te), y.mean())                       # baseline: the training average
    fitted[fname], test_pred[fname] = model, p

    m_rmse, b_rmse = rmse(y_te, p), rmse(y_te, b)
    boot_rmse = [rmse(y_te[i], p[i]) for i in idx_boot]
    boot_gain = [100 * (rmse(y_te[i], b[i]) - rmse(y_te[i], p[i])) / rmse(y_te[i], b[i]) for i in idx_boot]
    rng_lo, rng_hi = pre["regression"][fname]["expected_range_95"]
    rows.append({"features": fname, "model": info["model"],
                 "test RMSE": round(m_rmse, 3),
                 "RMSE 95% CI": f"{np.percentile(boot_rmse, 2.5):.2f} to {np.percentile(boot_rmse, 97.5):.2f}",
                 "test MAE": round(mean_absolute_error(y_te, p), 3),
                 "test R2": round(r2_score(y_te, p), 3),
                 "baseline RMSE": round(b_rmse, 3),
                 "gain over baseline %": round(100 * (b_rmse - m_rmse) / b_rmse, 1),
                 "gain 95% CI": f"{np.percentile(boot_gain, 2.5):.1f} to {np.percentile(boot_gain, 97.5):.1f}",
                 "inside expected range?": "Yes" if rng_lo <= m_rmse <= rng_hi else "NO"})

reg_test = pd.DataFrame(rows)
reg_test.to_csv("test_results_regression.csv", index=False)
reg_test
```

```python
# Cell 4: THE ONE-TIME EVALUATION, AT-RISK FLAGGING. The threshold was fixed in Cell 2.
rows, test_proba = [], {}
for fname, Xtr in feature_sets.items():
    c = cfg["classification"][fname]
    thr = pre["classification"][fname]["threshold"]
    model = build_clf(c["model"], c["params"]).fit(Xtr, y_risk)
    pr = model.predict_proba(test[Xtr.columns])[:, 1]
    test_proba[fname] = pr
    flag = (pr >= thr).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_te_risk, flag, labels=[0, 1]).ravel()

    aucs = [roc_auc_score(y_te_risk[i], pr[i]) for i in idx_boot if 0 < y_te_risk[i].sum() < len(i)]
    lo, hi = pre["classification"][fname]["expected_range_95"]
    auc = roc_auc_score(y_te_risk, pr)
    rows.append({"features": fname, "model": c["model"],
                 "test AUC": round(auc, 3),
                 "AUC 95% CI": f"{np.percentile(aucs, 2.5):.2f} to {np.percentile(aucs, 97.5):.2f}",
                 "PR AUC": round(average_precision_score(y_te_risk, pr), 3),
                 "PR AUC of guessing": round(y_te_risk.mean(), 3),
                 "threshold": round(thr, 2),
                 "flagged": int(flag.sum()), "caught": tp, "missed": fn, "false alarms": fp,
                 "recall": round(recall_score(y_te_risk, flag), 2),
                 "precision": round(precision_score(y_te_risk, flag, zero_division=0), 2),
                 "inside expected range?": "Yes" if lo <= auc <= hi else "NO"})

clf_test = pd.DataFrame(rows)
clf_test.to_csv("test_results_classification.csv", index=False)
open("test_evaluation_done.flag", "w").write(datetime.datetime.now().isoformat())
print("Test set evaluated. Do NOT rerun Cells 3 and 4.\n")
clf_test
```

```python
# Cell 5: look at the errors (a picture, not a new decision)
fig, axes = plt.subplots(2, 2, figsize=(12, 9))
for j, fname in enumerate(feature_sets):
    p = test_pred[fname]
    axes[0, j].scatter(p, y_te, alpha=0.6)
    axes[0, j].plot([4, 20], [4, 20], color="crimson", linestyle="--")
    axes[0, j].set_xlabel("Predicted G3"); axes[0, j].set_ylabel("Actual G3")
    axes[0, j].set_title(f"{fname}: actual vs predicted (test)")
    axes[1, j].hist(y_te - p, bins=15, color="steelblue", edgecolor="white")
    axes[1, j].axvline(0, color="crimson", linestyle="--")
    axes[1, j].set_xlabel("Residual (actual - predicted)"); axes[1, j].set_title("Test residuals")
plt.tight_layout()
plt.savefig("test_evaluation.png", dpi=150, bbox_inches="tight")
plt.show()

for fname in feature_sets:
    pe = test_pred[fname]
    low = y_te < 10
    print(f"{fname:18s} mean residual for at-risk students: {np.mean(y_te[low] - pe[low]):+.2f} "
          f"| for the others: {np.mean(y_te[~low] - pe[~low]):+.2f}")
```

```python
# Cell 6: cross-validated estimate vs test result, side by side
rows = []
for fname in feature_sets:
    r, pr_ = cfg["regression"][fname], pre["regression"][fname]
    t = reg_test[reg_test.features == fname].iloc[0]
    c, pc = cfg["classification"][fname], pre["classification"][fname]
    tc = clf_test[clf_test.features == fname].iloc[0]
    rows.append({"features": fname,
                 "CV RMSE (nested)": round(r["nested_cv_rmse"], 3), "test RMSE": t["test RMSE"],
                 "RMSE inside expected range": t["inside expected range?"],
                 "gain over baseline % (test)": t["gain over baseline %"], "gain 95% CI": t["gain 95% CI"],
                 "CV AUC (nested)": round(c["nested_roc_auc"], 3), "test AUC": tc["test AUC"],
                 "AUC inside expected range": tc["inside expected range?"]})
comparison = pd.DataFrame(rows)
comparison.to_csv("phase4_cv_vs_test.csv", index=False)
comparison.T
```

```python
# Cell 7: save the models and the information the app needs on Day 25
full = pd.concat([train, test], ignore_index=True)
y_full = full["G3"]
y_full_risk = (y_full < 10).astype(int)

meta = {"created": datetime.datetime.now().isoformat(timespec="seconds"),
        "target_regression": "G3 (final grade, 0-20)",
        "target_classification": "at risk = G3 < 10",
        "features_all": list(X_all.columns), "features_early": list(X_early.columns),
        "models": {}, "note": ("*_trainonly files are the models that were evaluated on the test set. "
                               "The other files are refitted on train+test with identical settings, for the app. "
                               "Reported performance refers to the train-only models.")}

for tag, fname in [("early", "early (no G1/G2)"), ("all", "all features")]:
    cols = list(feature_sets[fname].columns)
    r, c = cfg["regression"][fname], cfg["classification"][fname]

    joblib.dump(fitted[fname], f"reg_{tag}_trainonly.joblib")
    joblib.dump(build_reg(r["model"], r["params"]).fit(full[cols], y_full), f"reg_{tag}.joblib")

    clf_train = build_clf(c["model"], c["params"]).fit(feature_sets[fname], y_risk)
    joblib.dump(clf_train, f"clf_{tag}_trainonly.joblib")
    joblib.dump(build_clf(c["model"], c["params"]).fit(full[cols], y_full_risk), f"clf_{tag}.joblib")

    meta["models"][tag] = {"regression": r["model"], "classification": c["model"],
                           "threshold": pre["classification"][fname]["threshold"],
                           "test_rmse": float(reg_test[reg_test.features == fname]["test RMSE"].iloc[0]),
                           "test_auc": float(clf_test[clf_test.features == fname]["test AUC"].iloc[0]),
                           "regression_useful": r["useful"], "classification_usable": c["usable"]}

json.dump(meta, open("model_meta.json", "w"), indent=2)
print("Saved 8 model files, model_meta.json, test_results_*.csv, phase4_cv_vs_test.csv, test_evaluation.png")
print(json.dumps(meta["models"], indent=2))
```
